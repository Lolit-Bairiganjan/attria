```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print(f"Numpy version: {np.__version__}")
print(f"Pandas version: {pd.__version__}")
```

## Creating a raw_data_set variable for data wrangler

```python
raw_data_set = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")
raw_data_set
```

## Checking dimensions, data types and other important details of the data set

```python
print(f"Rows: {raw_data_set.shape[0]} \nColumns: {raw_data_set.shape[1]}]\n")
print(f"Data Types of columns:\n{'-'*len('Data Types of columns:')}\n{raw_data_set.dtypes}\n")
```

## Checking data with raw_data_set.info()

```python
print(raw_data_set.info())
```

## Checking for duplicates

```python
print(raw_data_set[raw_data_set['customerID'].duplicated()])
print(raw_data_set[raw_data_set['customerID'].duplicated()].sum())

# No duplicates found
```

## Dropping unnecessary columns

```python
raw_data_set = raw_data_set.drop(columns=['customerID'])
raw_data_set
```

## Non string columns in string to float/int

```python
raw_data_set['TotalCharges'] = pd.to_numeric(raw_data_set['TotalCharges'], errors='coerce')
print(raw_data_set.info())
```

## Mapping yes/no to 1/0

```python
yes_no_cols = ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn']

raw_data_set[yes_no_cols] = raw_data_set[yes_no_cols].apply(lambda x: x.map({
    "Yes": 1,
    "No": 0
}))

raw_data_set
```

## Mapping gender male/female to 1/0

```python
raw_data_set['gender'] = raw_data_set['gender'].map({
    'Male': 1,
    'Female': 0
})

raw_data_set
```

## Checking dataset again

```python
print(raw_data_set.info())
```

## Check plots to find corelations

```python
sns.countplot(data=raw_data_set, x='Churn')
plt.title('Churn Distribution')
plt.show()
```

```python
num_cols = raw_data_set.select_dtypes(include=['number'])

multival_cols = ['tenure', 'MonthlyCharges']
binary_cols = ['gender', 'SeniorCitizen', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling']

churned1_data = raw_data_set[raw_data_set['Churn'] == 1]
churned0_data = raw_data_set[raw_data_set['Churn'] == 0]
```

```python
for col in binary_cols:
    sns.countplot(data=churned1_data, x=col, color='red', alpha=0.5)
    sns.countplot(data=churned0_data, x=col, color='blue', alpha=0.5)
    plt.title(f"Distribution of {col} for churned customers")
    plt.show()
```

```python
for col in multival_cols:
    sns.histplot(data=churned1_data, x=col, color='red', alpha=0.5)
    sns.histplot(data=churned0_data, x=col, color='blue', alpha=0.5)
    plt.title(f"Distribution of {col} for churned customers")
    plt.show()
```
